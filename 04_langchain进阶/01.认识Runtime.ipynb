LangChain 的 create_agent 底层运行在 LangGraph Runtime 之上。Runtime 可以理解为 Agent 每次执行时拿到的“运行环境”，其中与数据管理最相关的三个部分是：
- Context：本次调用需要的静态身份和依赖。
- State：执行过程中不断变化的短期状态。
- Store：跨会话保存的长期数据。
