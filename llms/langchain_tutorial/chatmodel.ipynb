```python
from llms.langchain_tutorial.deepseek import get_deepseek_chat

chat = get_deepseek_chat()
```

# 阻塞模式

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="你是一个数学专家"),
    HumanMessage(content="什么是勾股定理"),
]

chat.invoke(messages)
```

# 流式模式

```python
for chunk in chat.stream(messages):
    print(chunk.content, end="|", flush=True)
```

# 缓存

```python
from langchain.globals import set_llm_cache
from langchain.cache import InMemoryCache
```

```python
%%time

set_llm_cache(InMemoryCache())

# The first time, it is not yet in cache, so it should take longer
chat.invoke("说一个笑话")
```

```python
%%time

# The second time it is, so it goes faster
chat.invoke("说一个笑话")
```

# DeepSeek 的 ChatModel 接口

```python
from llms.langchain_tutorial.deepseek import get_deepseek_chat

model = get_deepseek_chat()
```

```python
from langchain_core.messages import (
    AIMessage,
    BaseMessage,
    FunctionMessage,
    HumanMessage,
    SystemMessage,
    ToolMessage,
)

model.invoke(
    [
        HumanMessage(content="hello!"),
        AIMessage(content="Hi there human!"),
        HumanMessage(content="Meow!"),
    ]
)
```

```python
# 输入也支持字符串，可以等同于`[HumanMessage(content="cat vs dog")]`
for chunk in model.stream("cat vs dog"):
    print(chunk.content, end="|")
```

# 记录消耗tokens

```python
from langchain_core.messages import HumanMessage, SystemMessage
from llms.langchain_tutorial.callbacks.manager import get_generic_llms_callback

messages = [
        SystemMessage(content="你是一个数学专家"),
        HumanMessage(content="什么是勾股定理"),
    ]

with get_generic_llms_callback() as cb:
    chat.invoke(messages)
    print(cb)
```

```python
from llms.langchain_tutorial.deepseek import get_deepseek_chat

# 使用当前环境保存的 DeepSeek 密钥，或设置 DEEPSEEK_API_KEY
chat = get_deepseek_chat()

with get_generic_llms_callback() as cb:
    for chunk in chat.stream(messages):
        print(chunk.content, end="|", flush=True)
    
    print()
    print(cb)
```
